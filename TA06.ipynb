```python

import numpy as np
import matplotlib.pyplot as plt

# Parameter awal
N = 1000
S = 990
I = 10
R = 0

beta = 0.3
gamma = 0.1
hari = 100
dt = 1

# Menyimpan hasil simulasi
data_S = [S]
data_I = [I]
data_R = [R]
waktu = [0]

# Proses simulasi
for t in range(hari):
    dS = -beta * S * I / N
    dI = beta * S * I / N - gamma * I
    dR = gamma * I

    S = S + dS * dt
    I = I + dI * dt
    R = R + dR * dt

    data_S.append(S)
    data_I.append(I)
    data_R.append(R)
    waktu.append(t + 1)

# Membuat grafik
plt.figure(figsize=(10, 6))
plt.plot(waktu, data_S, label="Susceptible (S)")
plt.plot(waktu, data_I, label="Infected (I)")
plt.plot(waktu, data_R, label="Recovered (R)")

plt.xlabel("Hari")
plt.ylabel("Jumlah Populasi")
plt.title("Simulasi Penyebaran Penyakit Model SIR")
plt.legend()
plt.grid()
plt.show()
```
